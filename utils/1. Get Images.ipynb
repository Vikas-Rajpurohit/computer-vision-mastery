```python
import cv2


cap = cv2.VideoCapture(0)

num = 0

while cap.isOpened():

    succes, img = cap.read()
    img = cv2.flip(img,1)

    k = cv2.waitKey(5)

    if k == 27:
        break
    elif k == ord('s'): # wait for 's' key to save and exit
        cv2.imwrite('myImages/' + str(num) + '.png', img)
        print("image saved!")
        num += 1

    cv2.imshow('Image',img)
    
cap.release()
cv2.destroyAllWindows()
```
