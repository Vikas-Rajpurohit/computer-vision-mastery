## 1. Image Borders

```python
import cv2
import numpy as np

img = cv2.imread('Images/Image1.jpg')
img = cv2.resize(img, (400,600))

# Parameters = (src, Width (top, bottom, left, right), borderType, color)
brdr = cv2.copyMakeBorder(img, 2, 10, 6, 10, cv2.BORDER_CONSTANT, value =[255,255,125])

cv2.imshow('Border', brdr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

- Border types: OpenCV offers various border types, such as cv2.BORDER_CONSTANT, cv2.BORDER_REPLICATE, cv2.BORDER_REFLECT, and more.

## 2. Image Blending

```python
import cv2
import numpy as np

img1 = cv2.imread('Images/Image1.jpg')
img1 = cv2.resize(img1, (400,600))

img2 = cv2.imread('Images/Image2.jpg')
img2 = cv2.resize(img2, (400,600))

cv2.imshow('Image1', img1)
cv2.imshow('Image2', img2)

result1 = img2+img1 # Order doesn't matter
cv2.imshow('Result1', result1)

# Adding Image normally should be avoided
result2 = cv2.add(img2,img1) # Order doesn't matter
cv2.imshow('Result2', result2)

# Add weights for blending
result3 = cv2.addWeighted(img1,0.2,img2,0.8,0)
cv2.imshow('Result3', result3)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Plotting all Results using Matplotlib

```python
import matplotlib.pyplot as plt

def change(img):
    res = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return res

# Create a figure with multiple subplots
fig, axes = plt.subplots(nrows=2, ncols=3 ,figsize=(30,30))

# Plot data1 in the top-left subplot
axes[0, 0].imshow(change(img1))

# Plot data2 in the top-right subplot
axes[0, 2].imshow(change(img2))

# Plot data3 in the bottom-left subplot
axes[1, 0].imshow(change(result1))

# Plot data4 in the bottom-right subplot
axes[1, 1].imshow(change(result2))

# Plot data4 in the bottom-right subplot
axes[1, 2].imshow(change(result3))

# Set titles for each subplot
axes[0, 0].set_title('Plot 1')
axes[0, 1].set_title('Plot 2')
axes[1, 0].set_title('Plot 3')
axes[1, 1].set_title('Plot 4')

# Adjust the spacing between subplots
fig.tight_layout()

# Show the plots
plt.show()
```
