## 1. Import Statements

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 
import matplotlib.image as mpimg
```

```python
left_image = cv2.imread('Images/cone_left.png', cv2.IMREAD_GRAYSCALE)
right_image = cv2.imread('Images/cone_right.png', cv2.IMREAD_GRAYSCALE)
```

## 2. Disparity function

```python
stereo = cv2.StereoBM_create(numDisparities=0, blockSize=21)
# For each pixel algo will find the best disparity from 0
# Larger block size implies smoother, though less accurate disparity map
depth = stereo.compute(left_image, right_image)
```

## 3. Visulaization

```python
left = cv2.cvtColor(left_image, cv2.COLOR_BGR2RGB)
right = cv2.cvtColor(right_image, cv2.COLOR_BGR2RGB)

images = [left, right, depth]

fig, axs = plt.subplots(1, 3, figsize=(15, 5))

for i, image in enumerate(images):
    axs[i].imshow(image)
    axs[i].set_title(f'Image {i+1}')
    axs[i].axis('off')

# Adjust layout for better spacing
plt.tight_layout()

# Show the plot
plt.show()
```
